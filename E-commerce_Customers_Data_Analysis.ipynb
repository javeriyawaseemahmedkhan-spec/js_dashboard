```python
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(col, "=", len(outliers))
```

### Outliers Analysis Using IQR Method

The IQR method was used to identify outliers in all numerical columns.

The results show:

- Avg. Session Length: 3 outliers
- Time on App: 4 outliers
- Time on Website: 2 outliers
- Length of Membership: 12 outliers
- Yearly Amount Spent: 9 outliers

Length of Membership has the highest number of outliers, while Time on Website has the lowest number.

These values will be investigated before deciding whether to keep or remove them.

```python
col = "Length of Membership"

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df[(df[col] < lower_limit) | (df[col] > upper_limit)][col]
```

#Question 4:

Is there a relationship between Time on App and Yearly Amount Spent?

```python
sns.scatterplot(
    data=df,
    x="Time on App",
    y="Yearly Amount Spent"
)

plt.title("Time on App vs Yearly Amount Spent")
plt.xlabel("Time on App")
plt.ylabel("Yearly Amount Spent")
plt.show()
```

### Q4: Relationship Between Time on App and Yearly Amount Spent

The scatter plot shows a positive relationship between Time on App and Yearly Amount Spent.

Customers who spend more time on the mobile app generally tend to spend more money yearly.

This indicates that app usage is positively associated with yearly customer spending.

```python
df["Time on App"].corr(df["Yearly Amount Spent"])
```

### Correlation Analysis: Time on App and Yearly Amount Spent

The correlation coefficient between Time on App and Yearly Amount Spent is approximately 0.4993.

This indicates a moderate positive correlation between the two variables.

Customers who spend more time on the mobile app generally tend to have higher yearly spending.

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    df[["Time on App", "Time on Website",
        "Length of Membership", "Yearly Amount Spent"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

### Correlation Heatmap Interpretation

The heatmap shows the correlation between customer activity and yearly spending.

Length of Membership has the strongest positive correlation with Yearly Amount Spent (0.8091).

Time on App has a moderate positive correlation with Yearly Amount Spent (0.4993).

Time on Website has almost no linear correlation with Yearly Amount Spent (-0.0026).

This suggests that membership duration and mobile app usage are more strongly associated with yearly spending than website usage.
